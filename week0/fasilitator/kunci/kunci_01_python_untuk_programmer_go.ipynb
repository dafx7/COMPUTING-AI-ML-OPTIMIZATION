````python
import numpy as np

def cek(syarat, pesan="Belum tepat."):
    """Memeriksa jawabanmu. `syarat` adalah fungsi tanpa argumen yang mengembalikan True atau False."""
    try:
        benar = bool(syarat())
    except Exception as galat:
        benar = False
        pesan = f"{pesan} (Kode belum lengkap atau menimbulkan galat: {type(galat).__name__}: {galat})"
    print("Benar." if benar else "Belum tepat. " + pesan)
````

### Latihan 1: variabel, tipe, dan pembagian

**Go**
```go
total := 17
jumlahOrang := 4
rata := float64(total) / float64(jumlahOrang)
sisa := total % jumlahOrang
```
**Python** tidak memakai deklarasi tipe. Tiga hal yang perlu diingat:
- `/` selalu menghasilkan bilangan pecahan (float), meskipun kedua operannya bulat.
- `//` adalah pembagian bulat.
- `%` memberi sisa bagi, sama seperti di Go.

Isi tiga baris yang bertanda `...`.

````python
total = 17
jumlah_orang = 4
rata = total / jumlah_orang
hasil_bagi_bulat = total // jumlah_orang
sisa = total % jumlah_orang
````

````python
cek(lambda: rata == 4.25 and hasil_bagi_bulat == 4 and sisa == 1, 'Harapannya rata = 4.25, hasil bagi bulat = 4, sisa = 1.')
````

### Latihan 2: perulangan

**Go**
```go
jumlah := 0
for i := 1; i <= n; i++ {
    jumlah += i * i
}
```
**Python** memakai `for ... in range(...)`. Batas atas `range` tidak ikut dihitung, jadi `range(1, 4)` menghasilkan 1, 2, 3. Tidak ada kurung kurawal. Blok kode ditandai dengan **indentasi** (empat spasi).

Hitung jumlah kuadrat dari 1 sampai `n`.

````python
n = 10
jumlah = 0
for i in range(1, n + 1):
    jumlah += i * i
````

````python
cek(lambda: jumlah == 385, 'Jumlah kuadrat 1 sampai 10 seharusnya 385. Periksa batas atas range.')
````

### Latihan 3: list (padanan slice di Go)

**Go**
```go
data := []int{3, 1, 4, 1, 5, 9, 2, 6}
data = append(data, 7)
tiga := data[2:5]
panjang := len(data)
```
**Python**
- `list` adalah padanan slice. Tambah elemen dengan `.append(...)`, panjang dengan `len(...)`.
- Pengirisan `data[2:5]` bekerja sama seperti di Go: indeks 2, 3, 4.
- Python juga mengenal indeks negatif: `data[-1]` adalah elemen terakhir.

````python
data = [3, 1, 4, 1, 5, 9, 2, 6]
data.append(7)
tiga = data[2:5]
panjang = len(data)
terakhir = data[-1]
````

````python
cek(lambda: tiga == [4, 1, 5] and panjang == 9 and terakhir == 7 and data[-1] == 7, 'Setelah append, data punya 9 elemen. tiga harus [4, 1, 5].')
````

### Latihan 4: dict (padanan map di Go)

**Go**
```go
stok := map[string]int{"apel": 3, "jeruk": 0}
stok["pisang"] = 5
jumlah, ada := stok["mangga"]   // ada == false, jumlah == 0
```
**Python**
- `dict` ditulis dengan kurung kurawal: `{"apel": 3, "jeruk": 0}`.
- Menambah atau mengubah: `stok["pisang"] = 5`.
- Mengambil dengan nilai bawaan jika kunci tidak ada: `stok.get("mangga", 0)`.
- Memeriksa keberadaan kunci: `"mangga" in stok`.
- Menelusuri: `for nama, jumlah in stok.items():`.

````python
stok = {"apel": 3, "jeruk": 0}
stok["pisang"] = 5
mangga = stok.get("mangga", 0)
total_buah = 0
for nama, jumlah in stok.items():
    total_buah += jumlah
````

````python
cek(lambda: stok['pisang'] == 5 and mangga == 0 and total_buah == 8 and 'mangga' not in stok, 'Total buah seharusnya 3 + 0 + 5 = 8, dan stok tidak boleh berisi mangga.')
````

### Latihan 5: fungsi dengan dua nilai kembali

**Go**
```go
func minMaks(a []int) (int, int) {
    min, maks := a[0], a[0]
    for _, v := range a {
        if v < min { min = v }
        if v > maks { maks = v }
    }
    return min, maks
}
```
**Python**
```python
def nama_fungsi(parameter):
    ...
    return nilai_1, nilai_2     # sebenarnya mengembalikan satu tuple
```
Tidak ada tipe pada parameter, `if` tidak memakai kurung kurawal, dan perulangan `for v in a:` langsung memberi nilai elemen. Tulis fungsi `min_maks` tanpa memakai `min()` dan `max()` bawaan, supaya logikanya terlihat.

````python
def min_maks(a):
    kecil = a[0]
    besar = a[0]
    for v in a:
        if v < kecil:
            kecil = v
        if v > besar:
            besar = v
    return kecil, besar
````

````python
cek(lambda: min_maks([3, 1, 4, 1, 5, 9, 2, 6]) == (1, 9) and min_maks([-2, -7, -1]) == (-7, -1) and min_maks([5]) == (5, 5), 'Contoh: min_maks([3, 1, 4, 1, 5, 9, 2, 6]) harus (1, 9).')
````

### Latihan 6: penanganan galat

**Go**
```go
n, err := strconv.Atoi(teks)
if err != nil {
    return 0, err
}
```
**Python** tidak mengembalikan `err`. Operasi yang gagal **melempar pengecualian** (exception), dan kita menangkapnya dengan `try` dan `except`:
```python
try:
    n = int(teks)
except ValueError:
    ...
```
Tulis fungsi `ke_angka(teks)` yang mengembalikan bilangan bulat, atau `None` jika teks bukan bilangan bulat yang sah.

````python
def ke_angka(teks):
    try:
        return int(teks)
    except ValueError:
        return None
````

````python
cek(lambda: ke_angka('42') == 42 and ke_angka('abc') is None and ke_angka('3.5') is None, "ke_angka('42') harus 42, sedangkan 'abc' dan '3.5' harus None.")
````

### Latihan 8: indeks dan penyaringan

Array dua dimensi diindeks dengan `m[baris, kolom]`. Tanda `:` berarti "semua".
```python
m = np.array([[1, 2, 3],
              [4, 5, 6]])

m[1]           # baris kedua: array([4, 5, 6])
m[:, 0]        # kolom pertama: array([1, 4])

```

````python
m = np.array([[1, 2, 3],
              [4, 5, 6]])
kolom_terakhir = m[:, 2]
genap = m[m % 2 == 0]
````

````python
cek(lambda: list(kolom_terakhir) == [3, 6] and list(genap) == [2, 4, 6], 'kolom_terakhir harus [3, 6] dan genap harus [2, 4, 6].')
````

### Latihan 10: bilangan acak yang dapat diulang

Dalam optimasi dan ML kita sering memakai bilangan acak, tetapi percobaan harus **dapat diulang**. Caranya, buat pembangkit acak dengan `seed` tetap:
```python
rng = np.random.default_rng(42)
rng.uniform(-5, 5, size=3)    # tiga bilangan acak antara -5 dan 5
```
Dengan seed yang sama, urutan angkanya selalu sama. Buat pembangkit baru dengan seed 42, lalu ambil tiga titik.

````python
rng = np.random.default_rng(42)
titik = rng.uniform(-5, 5, size=3)
````

````python
cek(lambda: titik.shape == (3,) and bool(np.all((titik >= -5) & (titik <= 5))) and np.allclose(titik, np.random.default_rng(42).uniform(-5, 5, size=3)), 'Gunakan np.random.default_rng(42) lalu uniform(-5, 5, size=3) sebagai pemanggilan pertama.')
````

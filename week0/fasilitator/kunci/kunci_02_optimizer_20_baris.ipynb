Kamu akan mencari titik dengan **skor tertinggi** pada sebuah fungsi misterius f(x, y), tanpa boleh melihat rumusnya. Setiap kali kamu menanyakan skor satu titik, itu dihitung sebagai **satu evaluasi**. Anggaranmu hanya **30 evaluasi**.

Itu sudah seluruh masalahnya: sebuah **fungsi tujuan**, sebuah **ruang pencarian** (x dan y antara -5 dan 5), dan sebuah **anggaran**. Tiga istilah ini akan kita pakai sampai Pertemuan 16.

**Aturan**

- Jangan membagikan koordinat titikmu ke pasangan lain. Setiap pasangan mendapat peta yang berbeda.
- Memakai asisten AI diperbolehkan. Tetapi tuliskan dulu strategimu sendiri dalam satu kalimat, baru minta bantuan.

## 1. Persiapan
Ubah `KODE_PASANGAN` menjadi nomor yang diberikan mentor, lalu jalankan sel.

```python
import numpy as np
import matplotlib.pyplot as plt

KODE_PASANGAN = 0   # <- ganti dengan nomor pasangan dari mentor

class KotakHitam:
    """Fungsi misterius dengan anggaran evaluasi terbatas. Skor yang lebih tinggi lebih baik."""

    BATAS = 5.0

    def __init__(self, kode, anggaran=30):
        self.anggaran = anggaran
        self.riwayat = []                      # daftar (x, y, skor)
        gen = np.random.default_rng(kode)
        while True:
            self._pusat = gen.uniform(-self.BATAS + 1, self.BATAS - 1, size=(3, 2))
            jarak = [np.linalg.norm(self._pusat[i] - self._pusat[j])
                     for i in range(3) for j in range(i + 1, 3)]
            if min(jarak) > 3.0:
                break
        self._tinggi = np.array([1.0, 2.0, 3.0])
        self._lebar = np.array([2.0, 1.2, 0.9])
        kisi = np.linspace(-self.BATAS, self.BATAS, 201)
        gx, gy = np.meshgrid(kisi, kisi)
        self._optimum = self._nilai(gx, gy).max()

    def _nilai(self, x, y):
        x = np.asarray(x, dtype=float)
        y = np.asarray(y, dtype=float)
        total = 0.0
        for (cx, cy), t, w in zip(self._pusat, self._tinggi, self._lebar):
            total = total + t * np.exp(-((x - cx) ** 2 + (y - cy) ** 2) / (2 * w ** 2))
        return total

    def __call__(self, x, y):
        try:
            x = float(x)
            y = float(y)
        except (TypeError, ValueError):
            raise ValueError("x dan y harus berupa angka. Apakah bagian '...' sudah kamu isi?") from None
        if len(self.riwayat) >= self.anggaran:
            raise RuntimeError(f"Anggaran {self.anggaran} evaluasi sudah habis. "
                               "Buat KotakHitam baru untuk ronde berikutnya.")
        if abs(x) > self.BATAS or abs(y) > self.BATAS:
            raise ValueError("Titik di luar daerah. Pakai -5 sampai 5 untuk x dan untuk y.")
        skor = float(self._nilai(x, y))
        self.riwayat.append((x, y, skor))
        return skor

    def sisa(self):
        return self.anggaran - len(self.riwayat)

    def laporan(self):
        if not self.riwayat:
            print("Belum ada evaluasi.")
            return
        terbaik = max(r[2] for r in self.riwayat)
        persen = min(100.0, 100 * terbaik / self._optimum)
        print(f"Evaluasi terpakai : {len(self.riwayat)} dari {self.anggaran}")
        print(f"Skor terbaik      : {terbaik:.3f}  ({persen:.0f}% dari skor tertinggi yang mungkin)")

    def tunjukkan(self):
        kisi = np.linspace(-self.BATAS, self.BATAS, 200)
        gx, gy = np.meshgrid(kisi, kisi)
        fig, ax = plt.subplots(figsize=(6, 5))
        isi = ax.contourf(gx, gy, self._nilai(gx, gy), levels=20, cmap="viridis")
        fig.colorbar(isi, ax=ax, label="skor")
        for urutan, (x, y, s) in enumerate(self.riwayat, start=1):
            ax.plot(x, y, "o", color="white", markeredgecolor="black", markersize=7)
            ax.annotate(str(urutan), (x, y), textcoords="offset points", xytext=(4, 4), fontsize=7, color="white")
        if self.riwayat:
            bx, by, _ = max(self.riwayat, key=lambda r: r[2])
            ax.plot(bx, by, "*", color="#F7B125", markeredgecolor="black", markersize=16, label="titik terbaikmu")
            ax.legend(loc="upper right")
        ax.set_xlabel("x")
        ax.set_ylabel("y")
        ax.set_title("Peta skor dan titik yang kamu coba (angka = urutan)")
        plt.show()

print("Siap. Kode pasangan:", KODE_PASANGAN)
```

## 2. Ronde 1: pencarian acak

Strategi paling sederhana: coba 30 titik acak, ingat yang terbaik. Lengkapi bagian yang bertanda `...` dan `False`. Jumlahnya hanya beberapa baris.

```python
kotak = KotakHitam(KODE_PASANGAN)
rng = np.random.default_rng(1)

skor_terbaik = -1.0
titik_terbaik = None

for langkah in range(30):
    x = rng.uniform(-5, 5)
    y = rng.uniform(-5, 5)
    skor = kotak(x, y)
    if skor > skor_terbaik:
        skor_terbaik = skor
        titik_terbaik = (x, y)

print("Menurut catatanmu, skor terbaik:", round(skor_terbaik, 3))
kotak.laporan()
```

Tulis angka **persen**-mu di papan peringkat mentor, bersama kode pasanganmu.

## 3. Ronde 2: strategi bebas

Sekarang anggaran baru: 30 evaluasi pada peta yang sama. Hasil ronde 1 tidak ikut terhitung. Rancang strategi yang menurutmu lebih baik. Beberapa pertanyaan pemicu:
- Informasi apa dari evaluasi sebelumnya yang belum kamu pakai?
- Berapa evaluasi sebaiknya dipakai untuk **menjelajah**, dan berapa untuk **memperhalus**?
- Apa yang terjadi jika semua evaluasimu jatuh di sekitar satu bukit kecil?

```python
kotak2 = KotakHitam(KODE_PASANGAN)
rng2 = np.random.default_rng(2)

# Contoh strategi: jelajahi 15 titik acak, lalu perhalus di sekitar titik terbaik.
for _ in range(15):
    kotak2(rng2.uniform(-5, 5), rng2.uniform(-5, 5))
bx, by, bs = max(kotak2.riwayat, key=lambda r: r[2])
for _ in range(15):
    x = float(np.clip(bx + rng2.normal(0, 0.6), -5, 5))
    y = float(np.clip(by + rng2.normal(0, 0.6), -5, 5))
    s = kotak2(x, y)
    if s > bs:
        bx, by, bs = x, y, s

kotak2.laporan()
```

## 4. Lihat petanya
Setelah dua ronde selesai, tampilkan peta skor yang selama ini tersembunyi.

```python
kotak.tunjukkan()
kotak2.tunjukkan()
```
